```python
import pandas as pd
from pathlib import Path
import pyarrow.parquet as pq
```

### Read table


```python
# Define root folder
PROJECT_ROOT= Path.cwd().parent
# Khai báo đường dẫn liệu
DATA_DIR= PROJECT_ROOT/"data"/"raw"

PARQUET_PATH=DATA_DIR/"yellow_tripdata_2026-01.parquet"
LOOKUP_PATH=DATA_DIR/"taxi_zone_lookup.csv"

# Kiểm tra file tồn tại
print("Project root: ",PROJECT_ROOT)
print("Parquet exists: ",PARQUET_PATH.exists())
print("Lookup exists: ",LOOKUP_PATH.exists())
```

```python
df=pd.read_parquet(PARQUET_PATH)
zones= pd.read_csv(LOOKUP_PATH)
display(df.head(10))
display(zones.head(10))
```

### Kiểm tra cấu trúc dữ liệu

```python
print("Taxi data shape", df.shape)
print("Zone lookup shape",zones.shape)
```

```python
df.describe().T
```

Nhận định về dữ liệu ban đâu:

`passenger_count`: Min = 0, max = 9	Có giá trị 0 và số lượng hành khách bất thường cần kiểm tra

`trip_distance`: Min = 0, max = 269.097,48 Có chuyến 0 dặm và giá trị cực lớn; có thể ảnh hưởng trung bình

`fare_amount`L Min= -2555.2 max = 2.555,2 Có thể gồm giao dịch điều chỉnh hoặc hoàn tiền

`extra, mta_tax`: Có giá trị âm Cần kiểm tra quy tắc ghi nhận và giao dịch điều chỉnh

`tip_amount` :Min = -88,88; max = 766 Giá trị âm hoặc rất lớn cần kiểm tra

`tolls_amount` :Min = -94,5; max = 122,22 Cần xác định có phải điều chỉnh cước hay không

`total_amount` :Min = -2.560,2; max = 2.560,2 Cần kiểm tra tính hợp lý và cách tính tổng

`RatecodeID` :Max = 99 Có mã cần đối chiếu tài liệu dữ liệu

```python
# Kiểm tra giá trị null
showDFnull=df.isnull().sum()
showZNnull=zones.isnull().sum()
print(f"Giá trị null \n{showDFnull[showDFnull>0]}\n")
print(f"Giá trị Zones null \n{showZNnull[showZNnull>0]}")
```

```python
# Kiểm tra giá trị trùng lặp
showDFdup=df.duplicated().sum()
showZNdup=zones.duplicated().sum()
print(f"Giá trị trùng lặp taxi trip \n{showDFdup}\n")
print(f"Giá trị trùng lặp Zones \n{showZNdup}")
```
