```python
import pandas as pd
from pathlib import Path
import pyarrow.parquet as pq
```

### Read table


```python
# Define root folder
PROJECT_ROOT= Path.cwd().parent
# Khai báo đường dẫn liệu
DATA_DIR= PROJECT_ROOT/"data"/"raw"

PARQUET_PATH=DATA_DIR/"yellow_tripdata_2026-01.parquet"
LOOKUP_PATH=DATA_DIR/"taxi_zone_lookup.csv"

# Kiểm tra file tồn tại
print("Project root: ",PROJECT_ROOT)
print("Parquet exists: ",PARQUET_PATH.exists())
print("Lookup exists: ",LOOKUP_PATH.exists())
```

```python
# Kiểm tra giá trị null
showDFnull=df.isnull().sum()
showZNnull=zones.isnull().sum()
print(f"Giá trị null \n{showDFnull[showDFnull>0]}\n")
print(f"Giá trị Zones null \n{showZNnull[showZNnull>0]}")
```

```python
# Kiểm tra giá trị trùng lặp
showDFdup=df.duplicated().sum()
showZNdup=zones.duplicated().sum()
print(f"Giá trị trùng lặp taxi trip \n{showDFdup}\n")
print(f"Giá trị trùng lặp Zones \n{showZNdup}")
```

```python
# Kiểm tra các giá trị âm và <0 bất thường trong dữ liệu
check_issuseCols =[
    "passenger_count",
    "trip_distance",
    "fare_amount",
    "extra",
    "mta_tax",
    "tip_amount",
    "tolls_amount",
    "improvement_surcharge",
    "total_amount",
    "congestion_surcharge",
    "Airport_fee",
    "cbd_congestion_fee"
]

summary= pd.DataFrame({
    "zero_count": df[check_issuseCols].eq(0).sum(),
    "negative_count": df[check_issuseCols].lt(0).sum()
})

display(summary)
```

```python

# 1. Kiểm tra 10 chuyến có quãng đường lớn nhất
print("10 chuyến có trip_distance lớn nhất")

display(
    df.nlargest(10, "trip_distance")[
        [
            "tpep_pickup_datetime",
            "tpep_dropoff_datetime",
            "passenger_count",
            "trip_distance",
            "fare_amount",
            "total_amount"
        ]
    ]
)

# 2. Kiểm tra 10 chuyến có giá cước âm
print("10 chuyến có fare_amount âm nhỏ nhất")

display(
    df[df["fare_amount"] < 0].nsmallest(10, "fare_amount")[
        [
            "tpep_pickup_datetime",
            "tpep_dropoff_datetime",
            "passenger_count",
            "trip_distance",
            "fare_amount",
            "tip_amount",
            "total_amount"
        ]
    ]
)

# 3. Kiểm tra chuyến có quãng đường bằng 0
print("Số chuyến có trip_distance = 0:",
      df["trip_distance"].eq(0).sum())
```

```python

# Tính thời gian chuyến đi (phút)
duration = (
    df["tpep_dropoff_datetime"] - df["tpep_pickup_datetime"]
).dt.total_seconds() / 60

print("Kiểm tra thời gian chuyến đi")
print("Thời gian <= 0 phút:", (duration <= 0).sum())
print("Thời gian > 24 giờ:", (duration > 1440).sum())

display(
    duration.describe(
        percentiles=[0.25, 0.5, 0.75, 0.95, 0.99]
    )
)
```

```python
# Tính thời gian di chuyển
duration=(
    df['tpep_dropoff_datetime']-df['tpep_pickup_datetime']
).dt.total_seconds()/60

print("Kiểm tra thời gian chuyến đi")
print("Chuyến đi có thời gian <=0: ",(duration<0).sum())
print("Thời gian > 24 giờ:", (duration > 1440).sum())
```
