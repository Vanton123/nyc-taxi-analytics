```python
import pandas as pd
from pathlib import Path
from sqlalchemy import create_engine
```

```python
PROJECT_ROOT=Path.cwd().parent
DATA_DIR=PROJECT_ROOT/"data"/"processed"

trips=pd.read_parquet(DATA_DIR/"trip_clean.parquet")
zones=pd.read_csv(DATA_DIR/"taxi_zones.csv")

trips.columns.str.lower()
zones.columns.str.lower()

print(trips.shape, zones.shape)
print(trips.columns.tolist())
```

```python
import os
```

```python
from sqlalchemy.engine import URL
url = URL.create(
    drivername="postgresql+psycopg2",
    username="postgres",
    password=os.getenv("PG_PASSWORD", "MAT_KHAU"),,  
    host="localhost",
    port=5432,
    database="nyc_taxi",
)

engine=create_engine(url)

with engine.connect() as conn:
    print("Connect succesfully")
```

```python
# Nap data zones
zones.to_sql("taxi_zones", engine, if_exists="replace",index=False)
print("Completely")
```

```python
chunk_size = 100_000
for i in range(0, len(trips), chunk_size):
    chunk = trips.iloc[i:i + chunk_size]
    chunk.to_sql("trips_clean", engine,
                 if_exists="replace" if i == 0 else "append",
                 index=False, method="multi", chunksize=5000)
    print(f"Đã nạp {min(i + chunk_size, len(trips)):,} / {len(trips):,} dòng")
```
